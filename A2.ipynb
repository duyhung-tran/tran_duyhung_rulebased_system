```python
import random
from collections import defaultdict
```

```python
# Generative grammar for the 1st half of the story
sentence_structure = {
    'Sentence': [
        ['Character', 'Action', 'Location'],
        ['Character', 'Action', 'Location', 'Description'],
        ['Character', 'Interaction', 'Character', 'Location'],
        ['Character', 'Interaction', 'Character', 'Location', 'Description'],
        ['Character', 'Thought', 'Location'],
        ['Character', 'Thought', 'Location', 'Description'],
        ['Character', 'ObjectInteraction', 'Object', 'Location'],
        ['Character', 'ObjectInteraction', 'Object', 'Location', 'Description']
    ],
    'Character': [
        'Coraline',
        'The other mother',
        'The ghost children',
        'The black cat',
        'Mr. Bobinsky',
        'Wybie',
        'April',
        'Miriam'
    ],
    'Action': [
        'walked',
        'ran',
        'searched around',
        'sang',
        'found a small button',
        'ate a lot of food',
        'was frightened',
        'heard something strange'
    ],
    'Interaction': [
        'screamed at',
        'followed',
        'sew buttons in the eyes of',
        'fought',
        'laughed with',
        'spied on'
    ],
    'Location': [
        'next to the tiny door',
        "in Mr. Bobinsky's circus",
        "in April and Miriam's theatre",
        'in the garden',
        'in the Pink Palace',
        'in the hallway mirror',
        'in the snow globe'
    ],
    'Description': [
        'where everything looked the same as before',
        'where the food smells amazing',
        'where there are a lot of buttons',
        'where the music is so lively',
        'where the temperature is so low',
    ],
    'Thought': [
        'wondered why the eyes are buttons',
        'realized that something is off',
        'began to regret opening the tiny door',
        'wondered whether these people can be trusted',
    ],
    'Object': [
        'the key',
        'the tiny door',
        'the spider web',
        'the snow globe',
        'the eye button',
        'the circus rats',
        'the angel dogs',
        'the doll'
    ],
    'ObjectInteraction' : [
        'destroyed',
        'found',
        'threw',
        'looked at',
        'played with'
    ]
}

def generate(symbol, grammar):
    if isinstance(symbol, str):
        if symbol in sentence_structure:
            production = random.choice(sentence_structure[symbol])

            if isinstance(production, list):
                return ' '.join(
                    generate(sym, sentence_structure)
                    for sym in production
                )
            return production
    return symbol
```

```python
# Markov chain for the 2nd half of the story

markov_text = """
Coraline opened the tiny door and went into the other world.
Coraline talked with the ghost children in the hallway mirror.
Coraline enjoyed the good food made by the other mother.
The other father played piano and sang beautifully.
The other mother smiled and made good food.
The other mother transformed into a spider.
The other mother made the doll for Coraline.
The black cat followed Coraline to the other world.
The black cat helped Coraline to defeat the other mother.
The black cat has a strange voice.
April and Miriam loved to sing and perform.
April and Miriam lived in the Pink Palace's basement.
April and Miriam told Coraline her future is very dangerous.
Mr.Bobinsky and the rats are passionate about circus.
Mr.Bobinsky lived in the upper unit of the Pink Palace.
Mr.Bobinsky has a circus full of rats.
The ghost children are trapped inside the hallway mirror.
The ghost children are eaten by the other mother.
The ghost children used to live in the Pink Palace.
Wybie warned Coraline about the other mother.
Wybie rode a black bike.
Wybie and his grandma owned the Pink Palace.
"""

def build_markov_chain(text, order=2):
    chain = defaultdict(list)
    words = text.split()

    for i in range(len(words) - order):
        # Create state tuple from current words
        state = tuple(words[i:i+order])
        # Get the next word
        next_word = words[i+order]
        # Add to chain
        chain[state].append(next_word)

    return chain


def generate_text(chain, num_words=30):
    # Start with a random state from the chain
    words = list(random.choice(list(chain.keys())))
    order = len(list(chain.keys())[0])

    for _ in range(num_words - order):
        state = tuple(words[-order:])
        if state in chain:
            next_word = random.choice(chain[state])
            words.append(next_word)
        else:
            break

    return ' '.join(words)

# Build the chain
chain = build_markov_chain(markov_text)
```

```python
print("CORALINE: ANOTHER VERSION\n")


# First half: Generative Grammar
print("First half: Generative Grammar\n")
for i in range(8):
    print(generate('Sentence', sentence_structure) + ".")

print("\n")

# Second half: Markov Chain
print("Second half: Markov Chain\n")
generate_text(chain, num_words=70)
```
